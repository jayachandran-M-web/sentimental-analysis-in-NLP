```python
import pandas as pd
import numpy as np

# Text vectorization
from sklearn.feature_extraction.text import TfidfVectorizer

# Train-test split
from sklearn.model_selection import train_test_split

# Support Vector Machine classifier
from sklearn.svm import SVC

# Evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
reviews = pd.read_csv("review.csv", encoding='latin1', on_bad_lines='skip')

# Preview dataset
print(reviews.head())
```

```python
features = reviews['text']

# Label (dependent variable)
labels = reviews['sentiment']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    features,
    labels,
    test_size=0.2,
    stratify=labels,   # Keeps class distribution balanced
    random_state=42
)
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    max_features=5000,   # Limits vocabulary size
    ngram_range=(1, 2)   # Unigrams + Bigrams
)

# Fit on training data
X_train_tfidf = vectorizer.fit_transform(X_train)

# Transform test data
X_test_tfidf = vectorizer.transform(X_test)
```

```python
classifier = SVC(
    kernel='linear',  # Best for text classification
    C=1.0,
    random_state=42,
    probability=True
)

classifier.fit(X_train_tfidf, y_train)
```

```python
predictions = classifier.predict(X_test_tfidf)
```

```python
accuracy = accuracy_score(y_test, predictions)
print("Accuracy:", accuracy)

print("\nClassification Report:\n")
print(classification_report(y_test, predictions))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, predictions))
```

```python
def predict_sentiment(text):
    text_vector = vectorizer.transform([text])
    prediction = classifier.predict(text_vector)[0]
    probability = classifier.predict_proba(text_vector).max()
    return prediction, probability
```

```python
predict_sentiment("The course explanation was awful")
```
